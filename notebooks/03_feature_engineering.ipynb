```python
import pandas as pd
daily_data = pd.read_csv('../data/processed/onion_cleaned.csv', parse_dates=['Arrival_Date'])
daily_data = daily_data.sort_values(['Market', 'Arrival_Date']).reset_index(drop=True)
```

## **Lag Features**

```python
daily_data['Price_Lag_1'] = (daily_data.groupby('Market')['Modal_Price'].shift(1))
daily_data['Price_Lag_7'] = (daily_data.groupby('Market')['Modal_Price'].shift(7))
daily_data['Price_Lag_30'] = (daily_data.groupby('Market')['Modal_Price'].shift(30))
```

```python
daily_data.head(10)
```

## **Rolling averages**

```python
daily_data['Rolling_Mean_7'] = (
    daily_data
    .groupby('Market')['Modal_Price']
    .transform(lambda x: x.shift(1).rolling(window=7).mean())
)

daily_data['Rolling_Mean_30'] = (
    daily_data
    .groupby('Market')['Modal_Price']
    .transform(lambda x: x.shift(1).rolling(window=30).mean())
)
```

```python
daily_data.head(10)
```

## **seasonal indicator**

```python
daily_data['Month'] = daily_data['Arrival_Date'].dt.month
```

## **cross-market average price**

```python
market_avg_by_date = daily_data.groupby('Arrival_Date')['Modal_Price'].transform('mean')
```

```python
daily_data['Cross_Market_Avg'] = daily_data.groupby('Arrival_Date')['Modal_Price'].transform('mean')
```

```python
daily_data.isna().sum()
```

#### **Clean up leftover columns you don't need anymore**

```python
daily_data = daily_data.drop(columns=['Rolling_Median','Price_Ratio','Next_Price','Next_Rolling_Median','Next_Price_Ratio','Likely_Error'])
```

#### **dropping rows with NaN in actual feature columns**

```python
model_data = daily_data.dropna().reset_index(drop=True)
print(model_data.shape)
```

```python
model_data.to_csv('../data/processed/onion_model_ready.csv', index=False)
```
