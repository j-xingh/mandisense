```python
import pandas as pd
import numpy as np
model_data = pd.read_csv('../data/processed/onion_model_ready.csv', parse_dates=['Arrival_Date'])
model_data = model_data.sort_values(['Market', 'Arrival_Date']).reset_index(drop=True)
model_data.shape
```

```python
cutoff_date = '2023-01-01'
train = model_data[model_data['Arrival_Date'] < cutoff_date]
test = model_data[model_data['Arrival_Date'] >= cutoff_date]

print(train.shape, test.shape)
```

```python
feature_cols = ['Price_Lag_1', 'Price_Lag_7', 'Price_Lag_30', 'Rolling_Mean_7', 'Rolling_Mean_30', 'Month', 'Cross_Market_Avg']

X_train = train[feature_cols]
y_train = train['Modal_Price']
X_test = test[feature_cols]
y_test = test['Modal_Price']
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error

baseline_pred = X_test['Price_Lag_1']
baseline_mae = mean_absolute_error(y_test, baseline_pred)
print("Baseline MAE:", baseline_mae)
```

```python
print("y_test Mean:",y_test.mean())
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error

lr = LinearRegression()
lr.fit(X_train, y_train)

lr_pred = lr. predict(X_test)
lr_mae = mean_absolute_error(y_test, lr_pred)
print("Linear Regression MAE:", lr_mae)
```

```python
from xgboost import XGBRegressor

xgb = XGBRegressor(n_estimator=300, max_depth=5, learning_rate=0.05, random_state=42)
xgb.fit(X_train, y_train)

xgb_pred = xgb.predict(X_test)
xgb_mae = mean_absolute_error(y_test, xgb_pred)
print("XGBoost MAE:", xgb_mae)
```

```python
xgb2 = XGBRegressor(n_estimator=500, max_depth=7, learning_rate=0.03, random_state=42)
xgb2.fit(X_train, y_train)
xgb2_pred = xgb2.predict(X_test)
print("XGBoost (tuned) MAE:", mean_absolute_error(y_test, xgb2_pred))
```
