```python
import pandas as pd
import glob

files = glob.glob('../data/raw/*.csv')
print("Files found:", files)

dfs = []
for f in files:
    df = pd.read_csv(f)
    df['source_file'] = f
    dfs.append(df)

data = pd.concat(dfs, ignore_index=True)
print("Shape after concat:", data.shape)
print("Raw Arrival_Date sample:", data['Arrival_Date'].head(3).tolist())
print("Raw Arrival_Date sample (tail):", data['Arrival_Date'].tail(3).tolist())
```

```python
data['Arrival_Date'] = pd.to_datetime(data['Arrival_Date'], format='%d/%m/%Y', errors='coerce')
print("NaT count after parsing:", data['Arrival_Date'].isna().sum())
print("Max date after parsing:", data['Arrival_Date'].max())
```

```python
print("Grade value counts:")
print(data['Grade'].value_counts())

faq = data[data['Grade'] == 'FAQ']
print("\nShape after FAQ filter:", faq.shape)
print("Max date after FAQ filter:", faq['Arrival_Date'].max())
```

```python
market_counts = faq['Market'].value_counts()
print(market_counts.head(10))

locked_markets = market_counts.head(8).index.tolist()
print("\nLocked markets:", locked_markets)

filtered = faq[faq['Market'].isin(locked_markets)]
print("\nShape after market filter:", filtered.shape)
print("Max date after market filter:", filtered['Arrival_Date'].max())
print("\nPer-market max dates:")
print(filtered.groupby('Market')['Arrival_Date'].max())
```

```python
recent = faq[faq['Arrival_Date'] >= '2025-01-01']
recent_counts = recent['Market'].value_counts()
print(recent_counts.head(15))
```
